```python
import numpy as np
import matplotlip.pyplot as plt
from sklearn.datasets import load_breast_cancer
np.random.seed(42)
data=load_breast_cancer()
X,y=data.data,data.target
features_names=data.feature_names
target_names=data.target_names
print (f'dataset shape: x={X.shape} y={y.shape}')
```

```python
def train_test_split(X, y, test_size=0.2, seed=42):
    np.random.seed(seed)
    train_indices = []
    test_indices = []
    classes = np.unique(y)
    for c in classes:
        c_indices = np.where(y == c)[0]
        np.random.shuffle(c_indices)
        n_test = int(np.round(len(c_indices) * test_size))
        train_indices.extend(c_indices[:-n_test])
        test_indices.extend(c_indices[-n_test:])
    train_indices = np.array(train_indices)
    test_indices = np.array(test_indices)
    np.random.shuffle(train_indices)
    np.random.shuffle(test_indices)
    return X[train_indices], X[test_indices], y[train_indices], y[test_indices]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, seed=42)

def print_class_dis(name, y_vec):
    classes, counts = np.unique(y_vec, return_counts=True)
    total = len(y_vec)
    dis_str = ','.join([f'class {c}:{cnt}({cnt/total:.2%})' for c, cnt in zip(classes, counts)])
    print(f'{name} (Total={total})->{dis_str}')

print_class_dis('all data', y)
print_class_dis('train', y_train)
print_class_dis('test', y_test)

mean_train = np.mean(X_train, axis=0)
std_train = np.std(X_train, axis=0)
x_train_scaled = (X_train - mean_train) / std_train
x_test_scaled = (X_test - mean_train) / std_train
```

```python
def accuracy(y_true, y_pred):
    return np.mean(y_true == y_pred)

def confusion(y_true, y_pred, classes=None):
    if classes is None:
        classes = np.unique(np.concatenate([y_true, y_pred]))
    k = len(classes)
    cm = np.zeros((k, k), dtype=int)
    class_idx = {c: idx for idx, c in enumerate(classes)}
    for t, p in zip(y_true, y_pred):
        cm[class_idx[t], class_idx[p]] += 1
    return cm

def precision_f1(y_true, y_pred, classes=None):
    if classes is None:
        classes = np.unique(np.concatenate([y_true, y_pred]))
    cm = confusion(y_true, y_pred, classes)
    k = len(classes)
    precision = np.zeros(k)
    recalls = np.zeros(k)
    f1s = np.zeros(k)
    for i in range(k):
        tp = cm[i, i]
        fp = np.sum(cm[:, i]) - tp
        fn = np.sum(cm[i, :]) - tp
        precision[i] = tp / (tp + fp) if (tp + fp) > 0 else 0
        recalls[i] = tp / (tp + fn) if (tp + fn) > 0 else 0
        denom = precision[i] + recalls[i]
        f1s[i] = (2 * precision[i] * recalls[i] / denom) if denom > 0 else 0
    results = {
        'presion': precision,
        'recalls': recalls,
        'f1s': f1s,
        'macro_precision': np.mean(precision),
        'macro_recall': np.mean(recalls),
        'macro_f1': np.mean(f1s)
    }
    return results

def plot(cm, class_names, title="confusion matrix"):
    fig, ax = plt.subplots()
    im = ax.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
    ax.figure.colorbar(im, ax=ax)
    ax.set(xticks=np.arange(cm.shape[1]),
           yticks=np.arange(cm.shape[0]),
           xticklabels=class_names,
           yticklabels=class_names,
           title=title,
           ylabel='True label',
           xlabel='Predicted label')
    thresh = cm.max() / 2.0
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            ax.text(j, i, format(cm[i, j], 'd'),
                    ha='center', va='center',
                    color='white' if cm[i, j] > thresh else 'black', fontsize=12, fontweight='bold')
    plt.tight_layout()
    plt.show()

y_true_dummy = np.array([0, 0, 1, 1, 1])
y_pred_dummy = np.array([0, 0, 0, 0, 0])
print('dummy_accuracy:', accuracy(y_true_dummy, y_pred_dummy))
print('dummy_metrix(zero division check):', precision_f1(y_true_dummy, y_pred_dummy, classes=[0, 1]))
```

```python
def run_dep_ex(modelclass, x_train, y_train, x_test, y_test, fea_idx=20):
    copy_counts = [1, 3, 5, 10]
    acc = []
    avg_probs = []

    fea_train = x_train[:, [fea_idx]]
    fea_test = x_test[:, [fea_idx]]

    for c in copy_counts:
        if c == 1:
            x_tr_dup = x_train.copy()
            x_te_dup = x_test.copy()
        else:
            extra_tr = np.repeat(fea_train, c - 1, axis=1)
            extra_te = np.repeat(fea_test, c - 1, axis=1)
            x_tr_dup = np.hstack([x_train, extra_tr])
            x_te_dup = np.hstack([x_test, extra_te])

        clf = modelclass()
        clf.fit(x_tr_dup, y_train)
        preds = clf.predict(x_te_dup)
        probs = clf.predict_proba(x_te_dup)

        curr_acc = accuracy(y_test, preds)
        curr_avg = np.mean(np.max(probs, axis=1))

        acc.append(curr_acc)
        avg_probs.append(curr_avg)
        print(f'copies:{c:2d} | acc:{curr_acc:.4f} | avg_prob:{curr_avg:.4f}')

    fig, ax1 = plt.subplots(figsize=(8, 5))
    color1 = 'tab:blue'
    ax1.set_xlabel('number of copies')
    ax1.set_ylabel('test accuracy', color=color1)
    ax1.plot(copy_counts, acc, marker='o', color=color1, label='Accuracy')
    ax1.tick_params(axis='y', labelcolor=color1)
    ax1.set_ylim([0.85, 1.01])

    ax2 = ax1.twinx()
    color2 = 'tab:red'
    ax2.set_ylabel('average probability', color=color2)
    ax2.plot(copy_counts, avg_probs, marker='s', linestyle='--', color=color2, label='Avg Max Prob')
    ax2.tick_params(axis='y', labelcolor=color2)
    ax2.set_ylim([0.85, 1.01])

    plt.title(f"Effect of duplicating feature '{feature_names[fea_idx]}'")
    fig.tight_layout()
    plt.grid(True, alpha=0.3)
    plt.show()

    return acc, avg_probs
```
